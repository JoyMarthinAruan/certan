```python
# Perkalian dot product di vektor

a = np.array([1, 3])
b = np.array([2, 3])

perkalian_dot = np.dot(a, b)
print(perkalian_dot)
```

```python
# Invers matriks

a = np.array([(1, -1),(1, 1)])
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
# Solusi persamaan linear

A = np.array([(1, -6),(2, -5)])
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)

# Menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

# Cara lain
X2 = np.linalg.solve(A, Y)
print(X2)
```

```python
def f(x,y):
 return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))
x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)
X,Y = np.meshgrid(x,y)
Z = f(X,Y)
fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
#Tugas 2
x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)
X,Y = np.meshgrid(x,y)
Z = X**6 + 3*X**4*Y**2 - 3*X**4 + 3*X**2*Y**4 - X**2*Y**3 - 6*X**2*Y**2 + 3*X**2 + Y**6 - 3*Y**4 + 3*Y**2
plt.contour(X, Y, Z, levels = [1])
plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2 Dimensi')
plt.grid()
plt.show()
```

```python
#Tugas 3
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X,Y = np.meshgrid(x,y)
Z = ((5*Y/4) - np.abs(X))**2 + X**2
fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.plot_surface(X, Y, Z)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Grafik 3 Dimensi')
plt.show()
```
