##### penjelasan tahap 1 : mengimpor numpy untuk mengelola data angka dan Pandas untuk mengelola tabel.selanjutnya load_iris digunakan untuk menyimpannya dalam variabel iris

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Memuat dataset Iris
iris = load_iris()
```

##### penjelasan tahap 2 : membuat dan menambahkan kolom Species yang berisi jenis bunga.variabel x menyimpan empat fitur pengukuran bunga, sedangkan y menyimpan jenis bunga sebagai target prediksi.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Menentukan fitur dan target
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

##### penjelasan tahap 3 : untuk bagian ini menampilkan isi variabel y yaitu nama jenis bunga pada setiap baris data. Jenis bunga dalam dataset ini adalah setosa dan virginica

```python
y
```

##### penjelasan tahap 4 :mengubah nama jenis bunga menjadi angka menggunakan LabelEncoder.setosa diubah menjadi 0,versicolor menjadi 1 dan virginica menjadi 2.

```python
from sklearn.preprocessing import LabelEncoder

lab = LabelEncoder()
y = lab.fit_transform(y)

y
```

##### penjelasan tahap 5 : membagi dataset menjadi 80 persen data latih atau 120 data dan 20 persen data uji atau 30 data. 

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(
    X, test_size=0.2, random_state=0
)

y_train, y_test = train_test_split(
    y, test_size=0.2, random_state=0
)
```

##### penjelasan tahap 6 : membuat model Gaussian naïve bayes yang digunakan untuk fitur numerik kontinu,seperti ukuran bunga. Perintah fit() melatih model menggunakan data latih dan labelnya agar dapat memprediksi jenis bunga.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

##### penjelasan tahap 7 : memprediksi jenis bunga pada data uji menggunakan model yang sudah dilatih. Hasilnya disimpan dalam variabel result berupa angka 0,1,2 yang mewakili jenis bunga.

```python
result = NB_model.predict(X_test)

result
```

##### penjelasan tahap 8: membandingkan hasil prediksi dengan label sebenarnya untuk menghitung akurasi model. 

```python
from sklearn.metrics import accuracy_score

print(
    "Accuracy of data testing:",
    accuracy_score(y_test, result) * 100
)
```

##### penjelasan tahap 9 : memprediksi jenis bunga dari empat ukuran yang dimasukkan. Fungsi inverse_transform mengubah hasil prediksi angka menjadi nama bunga, yaitu setosa. Tahap ini dapat menampilkan peringatan karena input tidak memiliki nama kolom, sedangkan data pelatihan memiliki nama kolom.

```python
# Data baru: panjang sepal, lebar sepal,
# panjang petal, dan lebar petal
input_data = [[5.1, 3.5, 1.4, 0.2]]

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)

print("Hasil prediksi untuk input baru adalah:", prediction_label)
```

##### penjelasan tahap 10 : Kode ini mengubah input menjadi DataFrame dengan nama kolom yang sama seperti data pelatihan. Dengan begitu, model dapat memprediksi tanpa peringatan nama fitur. Hasil prediksinya tetap setosa.

```python
input_data = pd.DataFrame(
    np.array([[5.1, 3.5, 1.4, 0.2]]),
    columns=iris.feature_names
)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)

print("Hasil prediksi untuk input baru adalah:", prediction_label)
```

##### penjelasan tahap 11 : memasukkan data bunga pada soal ke dalam DataFrame, lalu memprediksi jenisnya menggunakan model Gaussian Naïve Bayes yang sudah dilatih. Hasil prediksi diubah dari angka menjadi nama spesies agar mudah dibaca. Urutan ukuran yang digunakan adalah panjang sepal, lebar sepal, panjang petal, dan lebar petal.

```python
# Data baru yang diberikan pada tugas
input_tugas = pd.DataFrame(
    [[7.3, 3.8, 4.6, 2.3]],
    columns=iris.feature_names
)

# Memprediksi jenis bunga
hasil_prediksi = NB_model.predict(input_tugas)

# Mengubah label angka menjadi nama spesies
nama_species = lab.inverse_transform(hasil_prediksi)

print("Data input:", input_tugas.values.tolist())
print("Hasil prediksi species:", nama_species[0])
```
